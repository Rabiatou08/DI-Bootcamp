Exercice 1 : Création et manipulation de tableaux Exercice 1 : Opérations matricielles

```python
import numpy as np

# Matrice 3x3
matrice = np.array([
    [1, 2, 3],
    [0, 1, 4],
    [5, 6, 0]
])

# Déterminant
det = np.linalg.det(matrice)

# Inverse
inverse = np.linalg.inv(matrice)

print("Matrice :")
print(matrice)

print("\nDéterminant :")
print(det)

print("\nInverse :")
print(inverse)
```

Exercice 5 : Représentation d'images

# Représentation des images avec NumPy

Les images sont représentées sous forme de tableaux NumPy.

- Pour une image en niveaux de gris, chaque pixel est représenté par une valeur comprise entre 0 (noir) et 255 (blanc).
- Une image 5x5 est donc un tableau de forme (5,5).
- Plus la valeur est élevée, plus le pixel est clair.

```python
import numpy as np

# Image 5x5 en niveaux de gris
image = np.array([
    [0, 50, 100, 150, 200],
    [25, 75, 125, 175, 225],
    [50, 100, 150, 200, 250],
    [75, 125, 175, 225, 255],
    [100, 150, 200, 225, 255]
])

print("Image 5x5 :")
print(image)

print("\nDimensions :", image.shape)

import matplotlib.pyplot as plt

plt.imshow(image, cmap='gray')
plt.title("Image 5x5 en niveaux de gris")
plt.show()
```

Exercice 6 : Tests d'hypothèses de base

```python
import numpy as np

# Productivité avant la formation
productivity_before = np.random.normal(loc=50, scale=10, size=30)

# Productivité après la formation
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

# Moyennes
mean_before = np.mean(productivity_before)
mean_after = np.mean(productivity_after)

# Écart-types
std_before = np.std(productivity_before)
std_after = np.std(productivity_after)

print("=== ANALYSE DE PRODUCTIVITÉ ===")
print(f"Moyenne avant formation : {mean_before:.2f}")
print(f"Moyenne après formation : {mean_after:.2f}")

print(f"\nÉcart-type avant : {std_before:.2f}")
print(f"Écart-type après : {std_after:.2f}")

# Différence moyenne
improvement = np.mean(productivity_after - productivity_before)

print(f"\nAmélioration moyenne : {improvement:.2f}")

# Conclusion simple
if mean_after > mean_before:
    print("\nConclusion : la productivité moyenne a augmenté après la formation.")
    print("Les données soutiennent l'hypothèse selon laquelle la formation est efficace.")
else:
    print("\nConclusion : aucune amélioration visible.")
```

### Hypothèse

H₀ : Le programme de formation n'a aucun effet sur la productivité.

H₁ : Le programme de formation améliore la productivité.

### Résultat

La moyenne de productivité après la formation est supérieure à celle observée avant la formation.

### Conclusion

Les données suggèrent que le programme de formation a eu un effet positif sur la productivité des employés.

 Exercice 9 : Conversion de données

```python
import numpy as np
import pandas as pd

# Tableau NumPy
array = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("Tableau NumPy :")
print(array)

# Conversion NumPy -> DataFrame Pandas
df = pd.DataFrame(
    array,
    columns=["A", "B", "C"]
)

print("\nDataFrame Pandas :")
print(df)

# Conversion DataFrame -> NumPy
array2 = df.to_numpy()

print("\nRetour en tableau NumPy :")
print(array2)
```
