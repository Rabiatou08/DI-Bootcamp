Défi quotidien : Scénarios d'analyse de données réell
 Défi : Alibaba et l'analyse de données

Présentation d'Alibaba
Alibaba est une plateforme e-commerce chinoise fondée en 1999
avec plus d'1 milliard d'utilisateurs dans le monde.

 Données collectées
- Historique de recherche
- Budget et prix préférés
- Localisation (pays)
- Préférences et taille
- Historique d'achats

 Méthodes utilisées
- Descriptive : analyser les habitudes d'achat
- Prédictive : anticiper les prochains achats
- Prescriptive : proposer les bons produits au bon moment

 Impact
Pour les utilisateurs :
- Trouvent facilement ce qu'ils cherchent
- Économisent du temps
- Découvrent de nouveaux produits

Pour Alibaba :
- Vendent plus de produits
- Fidélisent leurs clients
 Conclusion
Sans l'analyse de données, Alibaba serait une simple boutique
en ligne. Grâce aux données, ils personnalisent l'expérience
de chaque utilisateur et augmentent leurs ventes.
