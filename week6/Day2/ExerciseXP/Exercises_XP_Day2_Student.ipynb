## 🌟 Exercise 2: LLM Architecture and Application Scenarios
For each, describe (a) core architectural differences, (b) a real application, (c) why it fits.

### BERT
- Architecture: Bidirectional encoder-only model using Masked Language Modeling (MLM). It reads context from both left-to-right and right-to-left simultaneously.
- Application: Named Entity Recognition (NER) or Legal document search engines.
- Why: Because full context from both sides of a target token is required to resolve syntax ambiguities or detect legal entities accurately.

### GPT
- Architecture: Unidirectional decoder-only model using Causal Language Modeling (CLM). It hides future tokens using a causal mask, predicting the next word left-to-right.
- Application: Automated code generation assistants or conversational agents.
- Why: Autoregressive prediction allows it to synthesize fluid, creative, and completely novel sequences of symbols by relying strictly on the history of tokens generated so far.

### T5
- Architecture: Full Encoder-Decoder architecture unified under a flexible "text-to-text" operational framework.
- Application: Multi-document abstractive summarization and machine translation.
- Why: The encoder can fully digest the messy source sequence without direction constraints, while the decoder is left free to map out a totally transformed target string in a separate vocabulary space.


## 🌟 Exercise 3: Benefits and Ethics of Pre-training
 Benefits (explain each in your words):
1) Improved generalization: Upstream exposure to massive web text ensures the model develops global structural heuristics, avoiding overfitting on niche noise.

2) Less labeled data: Because the underlying syntax is pre-mastered, fine-tuning requires only a fraction of specialized annotated rows to learn the target task.

3) Faster fine-tuning: Models converge onto specialized objectives in very few epochs, saving compute expenses.

4) Transfer learning: Core semantic skills picked up from general web crawling successfully map directly onto separate expert fields (e.g., medical or financial text).

5) Robustness: The deep model gracefully absorbs typos, dynamic internet slang, and variations in word order without breaking down.


- Ethical concerns: Massive uncurated web data introduces institutional and historical human bias into the core weights. It can also propagate hallucinations (misinformation), facilitate automated malicious text creation (misuse), and inadvertently leak private individual information present in raw training datasets.

- Mitigations: Enforce strict initial dataset curation, apply differential privacy filters during training, run alignment cycles like RLHF (Reinforcement Learning from Human Feedback), deploy downstream toxicity barriers, and perform continuous external compliance audits.


## 🌟 Exercise 4: Transformer Architecture Deep Dive
### Self-Attention & Multi-Head Attention
- Q/K/V flow: Input tokens are mapped onto Query (Q), Key (K), and Value (V) matrices. Scaled dot-products between Q and K matrices measure token correlations, which are passed through a Softmax function to compute specific attention weights. These weights are then multiplied by the V matrix to aggregate contextual features across the sequence.

- Multiple heads: A single attention head can only evaluate one average projection. Splitting into multiple heads allows the network to project vectors into separate subspaces in parallel, capturing multiple semantic and syntactic perspectives simultaneously.

- Example sentence: "The bank approved the heavy industrial loan after reviewing the collateral."
- Head 1 could map the linguistic subject-object connection between "bank" and "approved".
- Head 2 could capture the financial context by binding the polysemic word "bank" to "loan" and "collateral", resolving the meaning of the institution rather than a river side.

### Pre-training Objectives
- MLM vs CLM: MLM conceals random interior tokens (e.g., [MASK]) and encourages the model to use left and right context to predict them. CLM hides all future positions, pushing the model to forecast the immediate next word.

- When to prefer: Prefer MLM for parsing tasks like classification or information extraction. Prefer CLM for generative tasks like writing assistants or code generation.

- NSP (Next Sentence Prediction) : Early BERT used NSP to help capture paragraph coherence. Modern models discard it because it was too simple a subtask, and extending continuous sequence lengths proved significantly better for long-range text coherence.

### Transformer Model Selection
- Sentiment on reviews: Encoder-only (e.g., BERT). Analyzing sentiment demands an exhaustive, bidirectional grasp of the text's tone and context without needing to generate new text.

- Conversational chatbot: Decoder-only (e.g., GPT). Chatbots require fluid, autoregressive text generation to extend dialogue sequences dynamically.

- Technical document translation: Encoder-Decoder (e.g., T5). This structure excels at completely mapping a whole source sentence (encoder) to a structurally separate target sentence (decoder).


### Positional Encoding
- Purpose: Since Transformers process all input tokens concurrently, they lose structural sequence sequence order. Positional encoding adds absolute or relative geometric coordinates (via sine/cosine waves) to vectors to inform the model where each word belongs.

- Failure example: Without position tracking, the sentences "The dog bit the cat" and "The cat bit the dog" would produce identical input embeddings, completely confounding the model regarding the agent of the action.

## 🌟 Exercise 6: Softmax Temperature: The Randomness Regulator
### 1) Temperature Scenarios
- T=0.2: The probability distribution sharpens aggressively onto the absolute top choice. Output becomes highly deterministic, conservative, repetitive, and factual.
- T=1.0: The model outputs tokens using the exact baseline probability distribution learned during training, maintaining an expected baseline of variance.
- T=1.5: The probability curve flattens significantly, lifting the odds of lower-ranked tokens. Output becomes highly chaotic, random, unpredictable, and prone to grammar errors or hallucinations.

### 2) Application Design
- Bedtime stories: Use a high temperature strategy around 0.7 to 0.9. This introduces imaginative plot lines, varied vocabularies, and surprising twists while preserving basic sentence syntax.
- Financial report summaries: Use a deterministic temperature strategy at 0.0 or 0.1. Creative liberties are unacceptable in finance; the model must strictly extract and summarize factual ledger details deterministically.

### 3) Temperature & Bias
Low temperatures restrict the model's output to the absolute highest-probability tokens, which correspond directly to the most dominant statistical archetypes in the training data. This accentuates biases by entirely eliminating alternative outcomes. For example, when completing "The nurse walked into the clinic and she..." at T=0.1, the model will output standard biased pronouns like "she" near 100% of the time. Increasing temperature to 0.8 allows the sampling space to open up to other plausible grammatical structures, breaking the structural bias loop.

```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
import torch

model_name = "gpt2"  # TODO: choose a causal LM
prompt = "Artificial intelligence will shape the future by"  # TODO: write your own prompt

pipe = pipeline(
    "text-generation",
    model=AutoModelForCausalLM.from_pretrained(model_name),
    tokenizer=AutoTokenizer.from_pretrained(model_name),
    device=0 if torch.cuda.is_available() else -1,
)

for T in [0.2, 1.0, 1.5]:
    out = pipe(prompt, max_new_tokens=40, temperature=T, do_sample=True)
    print(f"\n--- Temperature: {T}")
    print(out[0]["generated_text"])  # Inspect how style changes with T
```
