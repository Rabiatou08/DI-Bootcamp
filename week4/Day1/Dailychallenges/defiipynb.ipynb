## Défi : Comprendre l'essence du Machine Learning

### 1. Qu'est-ce que le Machine Learning ?

Le Machine Learning (ML) est une branche de l'intelligence artificielle
qui permet aux machines d'apprendre automatiquement à partir des données
sans être explicitement programmées.

**Importance pour les data analysts :**
- Automatiser les prédictions et décisions
- Découvrir des patterns cachés dans les données
- Traiter des volumes massifs de données rapidement
- Améliorer continuellement les performances avec plus de données

### 3. Les 3 types de Machine Learning

#### Apprentissage Supervisé
**Définition :** Le modèle apprend à partir de données étiquetées
(on connaît la bonne réponse).

**Types :**
- Classification → prédire une catégorie (oui/non, spam/pas spam)
- Régression → prédire une valeur numérique (prix, température)

**Exemple :** Prédire si un email est un spam ou non à partir
d'emails déjà étiquetés "spam" ou "non spam".

**Algorithmes :** Logistic Regression, Random Forest, XGBoost

---

#### Apprentissage Non Supervisé
**Définition :** Le modèle apprend à partir de données sans étiquettes.
Il découvre des structures cachées seul.

**Types :**
- Clustering → regrouper des données similaires
- Réduction de dimensionnalité → simplifier les données (PCA)

**Exemple :** Regrouper automatiquement des clients en segments
(jeunes, seniors, premium) sans catégories prédéfinies.

**Algorithmes :** K-Means, DBSCAN, PCA, t-SNE

---

#### Apprentissage par Renforcement
**Définition :** Un agent apprend par essais et erreurs en interagissant
avec un environnement. Il reçoit des récompenses ou pénalités.

**Exemple :** Un robot apprend à jouer aux échecs en jouant des milliers
de parties et en maximisant ses victoires.

**Algorithmes :** Q-Learning, Deep Q-Network (DQN), PPO

### 5. Processus de développement d'un modèle ML

#### Étape 1 — Sélection des caractéristiques (Feature Selection)
Choisir les variables les plus pertinentes pour le modèle.

**Méthodes :**
- Correlation Matrix → supprimer les variables corrélées
- Feature Importance → garder les variables les plus utiles
- PCA → réduire la dimensionnalité

**Pourquoi c'est important ?**
Trop de variables = modèle lent et imprécis (malédiction de la dimensionnalité)

---

#### Étape 2 — Sélection du modèle (Model Selection)
Choisir l'algorithme le plus adapté au problème.

**Critères de choix :**
- Type de problème (classification, régression, clustering)
- Taille du dataset (petit → simple, grand → complexe)
- Interprétabilité requise (banques → Decision Tree)
- Performance souhaitée (compétition → XGBoost, Neural Networks)

**Processus :**
1. Tester plusieurs modèles
2. Comparer les performances
3. Choisir le meilleur
4. Optimiser les hyperparamètres (GridSearchCV)

---

#### Étape 3 — Évaluation du modèle (Model Evaluation)
Mesurer les performances du modèle sur des données jamais vues.

**Métriques selon le type :**

Classification :
- Accuracy, Precision, Recall, F1-Score, AUC-ROC

Régression :
- MAE (Mean Absolute Error)
- RMSE (Root Mean Square Error)
- R² Score

Clustering :
- Score de Silhouette
- Méthode du coude (Elbow Method)

**Méthodes d'évaluation :**
- Train/Test Split (80/20)
- Cross-Validation (5 folds)
- Courbe ROC
