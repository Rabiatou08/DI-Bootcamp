## Exercice 1 : Importance de la visualisation

### Pourquoi la visualisation est importante ?
La visualisation des données permet de comprendre rapidement
des tendances, patterns et anomalies difficiles à voir dans
des tableaux de chiffres. Elle facilite la communication
des résultats à un public non technique.

### Utilité d'un graphique linéaire
Un graphique linéaire est utile pour visualiser l'évolution
d'une variable dans le temps (séries temporelles).
Exemple : température sur une semaine, ventes sur un an.

```python
# Exercice 2 : Graphique linéaire température
import matplotlib.pyplot as plt

jours = ['Lun', 'Mar', 'Mer', 'Jeu', 'Ven', 'Sam', 'Dim']
temperatures = [72, 74, 76, 80, 82, 78, 75]

plt.figure(figsize=(10, 5))
plt.plot(jours, temperatures, marker='o', color='blue', linewidth=2)
plt.xlabel('Jour')
plt.ylabel('Température (°F)')
plt.title('Variation de température sur une semaine')
plt.grid(True)
plt.show()
```

```python
# Exercice 3 : Graphique à barres ventes mensuelles
import matplotlib.pyplot as plt

mois = ['Jan', 'Fév', 'Mar', 'Avr', 'Mai']
ventes = [5000, 5500, 6200, 7000, 7500]

plt.figure(figsize=(10, 5))
plt.bar(mois, ventes, color='orange')
plt.xlabel('Mois')
plt.ylabel('Montant des ventes ($)')
plt.title('Ventes mensuelles du magasin')
plt.show()
```

```python
# Exercices 4, 5, 6 : Charger le dataset
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://github.com/devtlv/Datasets-GEN-AI-Bootcamp/raw/refs/heads/main/Week%203/W3D1%20-%20Data%20Visualization/Student%20Mental%20health.zip"
df = pd.read_csv(url, compression='zip')

display(df.head())
print(df.columns.tolist())
```

```python
# Exercice 4 : Visualisation de la distribution des moyennes cumulatives
import seaborn as sns
import matplotlib.pyplot as plt

# Voir les premières valeurs de la colonne CGPA
print(df['What is your CGPA?'].head())

# Créer l'histogramme
sns.histplot(
    data=df,
    x='What is your CGPA?',
    color='skyblue'
)

# Ajouter le titre
plt.title('Distribution des catégories de CGPA')

# Afficher le graphique
plt.show()
```

```python
# Exercice 5 : Anxiété selon le sexe
plt.figure(figsize=(10, 5))
sns.countplot(
    data=df,
    x='Do you have Anxiety?',
    hue='Choose your gender',
    palette='Set2'
)
plt.title('Niveaux d\'anxiété selon le sexe')
plt.xlabel('Anxiété')
plt.ylabel('Nombre d\'étudiants')
plt.show()
```

```python
# Exercice 6 : Âge vs crises de panique
df['Panic_numeric'] = df['Do you have Panic attack?'].map({'Yes': 1, 'No': 0})

plt.figure(figsize=(10, 5))
sns.scatterplot(
    data=df,
    x='Age',
    y='Panic_numeric',
    hue='Choose your gender',
    palette='Set1',
    s=100,
    alpha=0.7
)
plt.title('Relation entre âge et crises de panique')
plt.xlabel('Âge')
plt.ylabel('Crise de panique (1=Oui, 0=Non)')
plt.yticks([0, 1], ['Non', 'Oui'])
plt.show()
```
