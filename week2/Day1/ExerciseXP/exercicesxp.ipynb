## Exercice 7 : Transformation des données non structurées

| Source | Méthode de conversion | Justification |
|--------|----------------------|---------------|
| Articles de blog | NLP → extraire titre, date, mots-clés → CSV | Le texte doit être analysé avant stockage |
| Enregistrements audio | Speech-to-Text → extraire durée, sujet, sentiment → CSV | L'audio doit être converti en texte d'abord |
| Notes manuscrites | OCR → extraire les idées → Excel | Scanner + OCR pour lire le texte manuscrit |
| Tutoriel vidéo | Analyse vidéo → extraire titre, durée, étapes → CSV | La vidéo doit être analysée avant stockage |

Exercice 8 : Importer un fichier depuis Kaggle


```python
import pandas as pd

# Créer un DataFrame simple
df = pd.DataFrame({
    'Nom': ['Rabi', 'Ridi', 'Grace'],
    'Age': [22, 3, 20],
    'Ville': ['Abidjan', 'Yakro', 'Bouaké']
})

display(df)

# Exporter en Excel
df.to_excel('export.xlsx', index=False)
print(" Exporté en Excel !")

# Exporter en JSON
df.to_json('export.json')
print(" Exporté en JSON !")
```
