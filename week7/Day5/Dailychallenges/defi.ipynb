Scénario

Vous travaillez au sein de l'équipe Formation et Communication d'une multinationale du secteur technologique. On vous a demandé d'utiliser l'IA générative pour créer du contenu de micro-apprentissage destiné aux employés, pour une newsletter interne.

Votre tâche consiste à transformer un paragraphe d'une récente mise à jour de politique interne en un extrait concis et clair. Le texte original est trop formel, dense et truffé de jargon.

```python
# Initialiser l'API Gemini
gemini_model = genai.GenerativeModel('gemini-2.5-flash-preview-04-17')
```

### Étape 1 : Créer une invite qui contrôle la tonalité, le format et la durée de la sortie.

À partir de l'extrait de texte original suivant, rédigez une consigne qui :
* Le réécrit sur un ton amical et clair.
* Utilise des puces pour organiser les informations.
* Paraphrase le texte original, en évitant les citations directes.
* Le nombre total de mots reste inférieur à 75.

```python
input_text = """Les employés doivent s'assurer que tout accès distant aux systèmes internes est établi via le VPN sécurisé approuvé. En aucun cas, des connexions non sécurisées ou des appareils personnels dépourvus de protection des terminaux ne doivent être utilisés pour accéder à des données confidentielles ou à des communications sensibles."""

prompt = f"""Reformule le texte suivant sur un ton amical et clair. Utilise des puces pour organiser les informations. Paraphrase le texte original, en évitant les citations directes. La réponse doit contenir moins de 75 mots.

Texte :
{input_text}"""

response = gemini_model.generate_content(prompt)
print(response.text)
```

### Étape 3 : Détecter et atténuer les hallucinations

Si le modèle ajoute du contenu supplémentaire (par exemple, de nouveaux termes de politique, des technologies qui ne figurent pas dans l'entrée), rédigez une version révisée de votre invite originale qui limite strictement le modèle à l'entrée uniquement.

Par exemple, vous pourriez ajouter :

«Veuillez uniquement paraphraser le contenu du texte source. N’ajoutez pas de nouvelles recommandations ni de nouvelles technologies.»

```python
input_text = """Les employés doivent s'assurer que tout accès distant aux systèmes internes est établi via le VPN sécurisé approuvé. En aucun cas, des connexions non sécurisées ou des appareils personnels dépourvus de protection des terminaux ne doivent être utilisés pour accéder à des données confidentielles ou à des communications sensibles."""

# Modifiez le prompt pour atténuer les hallucinations en ajoutant une contrainte stricte.
prompt_step3 = f"""Reformule le texte suivant sur un ton amical et clair. Utilise des puces pour organiser les informations. Paraphrase le texte original, en évitant les citations directes. La réponse doit contenir moins de 75 mots.
Veuillez uniquement paraphraser le contenu du texte source. N’ajoutez pas de nouvelles recommandations ni de nouvelles technologies.

Texte :
{input_text}"""

response_step3 = gemini_model.generate_content(prompt_step3)
print(response_step3.text)
```

### Étape 4 : Analyse approfondie de la paraphrase

Rédigez maintenant une deuxième version de la consigne qui paraphrase le paragraphe de politique pour un public de stagiaires juniors, en utilisant un langage simple et des phrases courtes.

* Utilisez au maximum 4 puces
* Évitez tout jargon d'entreprise ou juridique.
* Adoptez un ton encourageant et informatif.

```python
input_text = """Les employés doivent s'assurer que tout accès distant aux systèmes internes est établi via le VPN sécurisé approuvé. En aucun cas, des connexions non sécurisées ou des appareils personnels dépourvus de protection des terminaux ne doivent être utilisés pour accéder à des données confidentielles ou à des communications sensibles."""

# Prompt pour paraphraser pour un public de stagiaires juniors
prompt_step4 = f"""Paraphrase le texte suivant pour un public de stagiaires juniors. Utilise un langage simple, des phrases courtes et un ton encourageant et informatif. La réponse doit contenir au maximum 4 puces et éviter tout jargon d'entreprise ou juridique.

Texte :
{input_text}"""

response_step4 = gemini_model.generate_content(prompt_step4)
print(response_step4.text)
```

### Étape 5 : Variante d'extraction de citation

Parfois, citer est plus efficace que paraphraser. Rédigez une nouvelle consigne qui extrait une citation directe du texte original qui résume le mieux la politique de sécurité essentielle.

Répondez ensuite :

* Dans quel type de communication interne la citation serait-elle plus appropriée que la paraphrase ?
* Dans quelles circonstances les devis peuvent-ils présenter un risque ?

```python
input_text = """Les employés doivent s'assurer que tout accès distant aux systèmes internes est établi via le VPN sécurisé approuvé. En aucun cas, des connexions non sécurisées ou des appareils personnels dépourvus de protection des terminaux ne doivent être utilisés pour accéder à des données confidentielles ou à des communications sensibles."""

# Prompt pour extraire une citation directe
prompt_step5 = f"""Extrayez une citation directe du texte suivant qui résume le mieux la politique de sécurité essentielle.

Texte :
{input_text}"""

response_step5 = gemini_model.generate_content(prompt_step5)
print(response_step5.text)

# Réponse aux questions :
print("\n--- Questions --- ")
print("1. Dans quel type de communication interne la citation serait-elle plus appropriée que la paraphrase ?")
print("   Réponse : Une citation directe est plus appropriée dans les communications formelles, les documents de politique officiels, les formations juridiques ou de conformité, ou lorsque la précision et l'autorité du texte original sont primordiales. Elle assure qu'aucun sens n'est perdu dans la reformulation.")
print("\n2. Dans quelles circonstances les devis peuvent-ils présenter un risque ?")
print("   Réponse : Les citations peuvent présenter un risque si le texte original est trop technique, jargonnesque ou difficile à comprendre pour le public cible, ce qui pourrait entraîner des malentendus. Elles peuvent aussi être trop longues ou ne pas s'intégrer naturellement dans le flux d'une communication plus informelle. De plus, une citation hors contexte peut être mal interprétée ou utilisée à mauvais escient.")
```
