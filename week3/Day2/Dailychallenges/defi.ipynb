```python
import pandas as pd

df = pd.read_csv('global_power_plant_database.csv')

print(df.head())
```

```python
print(df.shape)
print(df.columns.tolist())
df.info()
```

```python
# Valeurs manquantes
print(df.isnull().sum())

# Doublons
print("Doublons :", df.duplicated().sum())

# Suppression des doublons
df = df.drop_duplicates()

print(df.describe())
```

```python
country_counts = df['country'].value_counts()

print(country_counts.head(10))
```

```python
fuel_counts = df['primary_fuel'].value_counts()

print(fuel_counts)
```

```python
fuel_capacity = df.groupby('primary_fuel')['capacity_mw'].mean()

print(fuel_capacity.sort_values(ascending=False))
```

```python
print(df.columns.tolist())
```

```python
print(df.isnull().sum().sort_values(ascending=False))

df['capacity_mw'] = pd.to_numeric(df['capacity_mw'], errors='coerce')
df['commissioning_year'] = pd.to_numeric(df['commissioning_year'], errors='coerce')
```

```python
import numpy as np

print("Capacité moyenne :", np.mean(df['capacity_mw']))
print("Capacité médiane :", np.median(df['capacity_mw']))
print("Écart-type :", np.std(df['capacity_mw']))
```

```python
fuel_count = df['primary_fuel'].value_counts()

print(fuel_count)
```

```python
import matplotlib.pyplot as plt

fuel_count.head(10).plot(kind='bar')

plt.title('Power Plants by Fuel Type')
plt.xlabel('Fuel Type')
plt.ylabel('Number of Plants')
plt.xticks(rotation=45)

plt.show()
```

```python
country_count = df['country_long'].value_counts().head(10)

print(country_count)

country_count.plot(kind='bar')

plt.title('Top 10 Countries by Number of Power Plants')
plt.xlabel('Country')
plt.ylabel('Number of Plants')

plt.show()
```

```python
fuel_capacity = (
    df.groupby('primary_fuel')['capacity_mw']
      .mean()
      .sort_values(ascending=False)
)

print(fuel_capacity)

fuel_capacity.plot(kind='bar')

plt.title('Average Capacity by Fuel Type')
plt.ylabel('Capacity (MW)')
plt.xlabel('Fuel Type')

plt.show()
```

```python
year_data = (
    df['commissioning_year']
    .dropna()
    .astype(int)
    .value_counts()
    .sort_index()
)

plt.figure(figsize=(12,6))

plt.plot(year_data.index, year_data.values)

plt.title('Power Plants Commissioned Over Time')
plt.xlabel('Year')
plt.ylabel('Number of Plants')

plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.scatter(
    df['longitude'],
    df['latitude'],
    alpha=0.3
)

plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.title('Global Distribution of Power Plants')

plt.show()
```

```python
matrix = df[
    ['capacity_mw','latitude','longitude']
].dropna().head(100).to_numpy()

print(matrix.shape)
corr_matrix = np.corrcoef(matrix.T)

print(corr_matrix)
eigenvalues, eigenvectors = np.linalg.eig(corr_matrix)

print("Eigenvalues :")
print(eigenvalues)

print("\nEigenvectors :")
print(eigenvectors)
```

```python
top_country = df['country_long'].value_counts().idxmax()

top_fuel = df['primary_fuel'].value_counts().idxmax()

largest_capacity_fuel = (
    df.groupby('primary_fuel')['capacity_mw']
      .mean()
      .idxmax()
)

print("=== EXECUTIVE SUMMARY ===")

print(f"\nCountry with most power plants: {top_country}")

print(f"\nMost common fuel type: {top_fuel}")

print(f"\nFuel type with highest average capacity: {largest_capacity_fuel}")

print("\nKey Findings:")
print("- Renewable and fossil fuel plants dominate the dataset.")
print("- Power plant construction increased significantly in recent decades.")
print("- Capacity varies strongly according to fuel type.")
print("- Geographic distribution shows strong concentration in industrialized countries.")
```
