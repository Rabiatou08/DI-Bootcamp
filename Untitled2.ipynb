L'analyse de données est le processus de collecte, nettoyage

## Exercice 2 : Chargement et analyse initiale

```python
import pandas as pd
import zipfile

# Dataset 1 : Sommeil
df1 = pd.read_csv('How Much Sleep Do Americans.zip', compression='zip')
display(df1.head())
print(df1.shape)
df1.info()
display(df1.describe())

# Dataset 2 : Santé mentale
df2 = pd.read_csv('Global Trends in Mental Health Disorder.zip', compression='zip')
display(df2.head())
print(df2.shape)
df2.info()
display(df2.describe())

# Dataset 3 : Cartes de crédit
with zipfile.ZipFile('Credit Card Approvals (Clean Data).zip') as z:
    df3 = pd.read_csv(z.open('Credit Card Approvals (Clean Data)/clean_dataset.csv'))
display(df3.head())
print(df3.shape)
df3.info()
display(df3.describe())
```


### Description Dataset 1 : Sommeil des Américains

- Données de 2003 à 2017
- Habitudes de sommeil par âge, sexe et année
- Moyenne de sommeil : 8.8h par jour

### Description Dataset 2 : Santé mentale mondiale

### Description Dataset 3 : Cartes de crédit

## Exercice 7 : Transformation des données non structurées

| Source               | Méthode de conversion                              | Justification                          |
|----------------------|----------------------------------------------------|----------------------------------------|
| Articles de blog     | NLP → extraire titre, date, mots-clés → CSV        | Le texte doit être analysé avant stockage |
| Enregistrements audio| Speech-to-Text → extraire durée, sujet → CSV       | L'audio doit être converti en texte    |
| Notes manuscrites    | OCR → extraire les idées → Excel                   | Scanner + OCR pour lire le manuscrit   |
| Tutoriel vidéo       | Analyse vidéo → extraire titre, étapes → CSV       | La vidéo doit être analysée avant stockage |
