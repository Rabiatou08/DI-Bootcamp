Défi quotidien : Intégration de NumPy, Pandas et Matplotlib

```python
import numpy as np
import pandas as pd

villes = [
    "Abidjan", "Paris", "Londres", "New York", "Tokyo",
    "Dakar", "Nairobi", "Le Caire", "Sydney", "Rio"
]

mois = [
    "Jan", "Fev", "Mar", "Avr", "Mai", "Juin",
    "Juil", "Aout", "Sep", "Oct", "Nov", "Dec"
]
temp = np.random.uniform(-5, 35, (10, 12))
print(temp.shape)

data = pd.DataFrame(
    temp,
    index=villes,
    columns=mois
)

print(data)

moyenne_ville = data.mean(axis=1)

print(moyenne_ville)

ville_plus_chaude = moyenne_ville.idxmax()

print(ville_plus_chaude)

ville_plus_froide = moyenne_ville.idxmin()

print(ville_plus_froide)

print("Ville la plus chaude :", ville_plus_chaude)
print("Ville la plus froide :", ville_plus_froide)

print("Température moyenne :")
print(moyenne_ville)
```

```python
import matplotlib.pyplot as plt

moyenne_ville.plot(kind='bar')

plt.title("Température moyenne annuelle par ville")
plt.xlabel("Villes")
plt.ylabel("Température (°C)")

plt.show()
```

```python
data.T.plot(figsize=(10,6))

plt.title("Evolution mensuelle des températures")
plt.xlabel("Mois")
plt.ylabel("Température (°C)")

plt.show()

print("\n===== RAPPORT =====")

print(
    f"La ville la plus chaude est {ville_plus_chaude} "
    f"avec une température moyenne de {moyenne_ville.max():.2f} °C."
)

print(
    f"La ville la plus froide est {ville_plus_froide} "
    f"avec une température moyenne de {moyenne_ville.min():.2f} °C."
)
```
