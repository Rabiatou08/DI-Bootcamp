Défi quotidien : Gestion et analyse des données en Python

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.decomposition import PCA

# Charger le dataset depuis GitHub
url = "https://github.com/devtlv/Datasets-DA-Bootcamp-2-/raw/refs/heads/main/Week%204%20-%20Data%20Understanding/W4D4%20-%20Data%20Preprocessing%20%26%20T/Data%20Science%20Job%20Salary%20dataset.zip"
df = pd.read_csv(url, compression='zip')

display(df.head())
print(df.shape)
print(df.columns.tolist())

# Normalisation Min-Max
scaler = MinMaxScaler()
df['salary_normalized'] = scaler.fit_transform(df[['salary']])
print("\n=== Salary normalized ===")
print(df[['salary', 'salary_normalized']].head())
print(f"\nMin : {df['salary_normalized'].min():.2f}")
print(f"Max : {df['salary_normalized'].max():.2f}")

# PCA sur colonnes numériques
numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns
pca = PCA(n_components=2)
df_pca = pca.fit_transform(df[numeric_cols])
print("\nPCA shape :", df_pca.shape)

# Agrégation
result = df.groupby('experience_level')['salary'].agg(['mean', 'median'])
print("\n=== Salaire par niveau d'expérience ===")
print(result)
```
