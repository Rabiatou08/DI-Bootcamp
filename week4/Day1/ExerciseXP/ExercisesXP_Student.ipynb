## 🌟 Exercise 3 : Training, Evaluating, and Optimizing the Model

### Instructions
Which model(s) would you pick for a Loan Prediction ?
Outline the steps to evaluate the model’s performance, mentioning specific metrics that would be relevant to evaluate the model.

### Guidance
- Begin by naming one or two candidate models in a complete sentence and explain why each model is suitable for this problem.

```python

# This piece of code is already prefilled, run it to execute it and see the results.
# It demonstrates standard classification metrics for binary loan default prediction.

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, average_precision_score, confusion_matrix, classification_report

# Please replace these placeholders with your true labels and predicted probabilities.
y_true = [0,1,0,1,0,0,1,0,1,0]            # placeholder labels
y_pred_proba = [0.05,0.80,0.10,0.65,0.20,0.15,0.70,0.30,0.85,0.25]  # placeholder probabilities

# You should set a decision threshold that reflects the precision–recall trade-off for your business case.
threshold = 0.5
y_pred = [1 if p >= threshold else 0 for p in y_pred_proba]

print("Accuracy:", round(accuracy_score(y_true, y_pred), 4))
print("Precision:", round(precision_score(y_true, y_pred, zero_division=0), 4))
print("Recall:", round(recall_score(y_true, y_pred, zero_division=0), 4))
print("F1-score:", round(f1_score(y_true, y_pred, zero_division=0), 4))
print("ROC-AUC:", round(roc_auc_score(y_true, y_pred_proba), 4))
print("PR-AUC (Average Precision):", round(average_precision_score(y_true, y_pred_proba), 4))
print("\nConfusion matrix:\n", confusion_matrix(y_true, y_pred))
print("\nClassification report:\n", classification_report(y_true, y_pred, zero_division=0))
```

## Exercice 5 : Stratégie d'évaluation pour différents modèles ML

---

### 1. Modèle Supervisé — Classification (ex: prédiction de prêt)

#### Métriques
| Métrique | Description | Quand l'utiliser |
|---|---|---|
| Accuracy | % bonnes prédictions | Données équilibrées |
| Precision | Éviter faux positifs | Coût élevé des FP |
| Recall | Éviter faux négatifs | Coût élevé des FN |
| F1-Score | Équilibre P et R | Données déséquilibrées |
| AUC-ROC | Performance globale | Vue d'ensemble |

#### Méthodes
- **Train/Test Split** → 80% entraînement / 20% test
- **Cross-Validation (5 folds)** → évaluation plus robuste
- **Courbe ROC** → visualiser le compromis FP/FN
- **Matrice de confusion** → détailler les erreurs

#### Difficultés
- Données déséquilibrées (ex: 95% approuvés, 5% refusés)
- Overfitting si le modèle mémorise les données d'entraînement
- Choix de la bonne métrique selon le contexte métier

---

### 2. Modèle Non Supervisé — Clustering (ex: bibliothèque)

#### Métriques
| Métrique | Description |
|---|---|
| Score de Silhouette | Mesure la qualité des clusters (-1 à 1) |
| Inertie (Elbow Method) | Trouver le bon nombre de clusters K |
| Davies-Bouldin Index | Plus petit = meilleurs clusters |

#### Méthodes
- **Méthode du coude** → tracer l'inertie selon K pour trouver le K optimal
- **Score de Silhouette** → proche de 1 = bien séparé, proche de 0 = chevauchement
- **Visualisation PCA** → réduire à 2D pour visualiser les clusters

#### Difficultés
- Pas de "bonne réponse" pour comparer
- Résultats différents selon l'initialisation
- Difficile de choisir le bon nombre de clusters K

---

### 3. Modèle Renforcement — Robot dans un labyrinthe

#### Métriques
| Métrique | Description |
|---|---|
| Récompense cumulative | Somme des récompenses sur un épisode |
| Convergence | Le modèle cesse d'améliorer sa politique |
| Taux de succès | % fois où le robot atteint l'objectif |
| Nombre de pas | Moins de pas = chemin plus court |

#### Méthodes
- **Courbe de récompense** → tracer la récompense par épisode
- **Epsilon-Greedy** → équilibre exploration (tester) / exploitation (utiliser)
- **Taux de convergence** → combien d'épisodes avant stabilisation

#### Difficultés
- Equilibre exploration/exploitation difficile à trouver
- Récompenses rares → apprentissage très lent
- Environnement instable → résultats non reproductibles

---

### Résumé comparatif

| Type de ML | Métrique principale | Difficulté principale |
|---|---|---|
| Supervisé | F1-Score, AUC-ROC | Overfitting, données déséquilibrées |
| Non supervisé | Score de Silhouette | Pas de vérité terrain |
| Renforcement | Récompense cumulative | Exploration vs exploitation |
