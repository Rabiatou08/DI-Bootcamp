## Exercise 1: Open Source Levels Reflection
**Use:** Fully Open, Weights Released, Architecture Only; model components; fine-tuning/domain adaptation; healthcare compliance.

**Step-by-step**
1) Gather definitions for the three openness levels.  
2) Identify key characteristics (what is open; what you can/cannot do).  
3) Compare side by side in a table (what's open vs. impact).  
4) Write a 3?5 sentence comparative paragraph.  
5) Answer the healthcare prompt (1?2 sentences) about retraining on clinical data.

**Deliverables:** paragraph + healthcare answer.

```python
open_source_levels = {
    "Fully Open": {
        "definition": "All components, including training data, model architecture, weights, and training code, are publicly available and can be used, modified, and distributed without significant restrictions.",
        "what_is_open": ["Training data", "Model architecture", "Model weights", "Training code", "Evaluation code", "Usage policies"],
        "what_you_can_do": ["Replicate training", "Fine-tune extensively", "Adapt for specific domains", "Inspect for biases", "Use commercially (depending on license)", "Research and develop new techniques"],
        "what_you_cannot_do": ["Few restrictions beyond standard license compliance (e.g., responsible use, attribution)"],
    },
    "Weights Released": {
        "definition": "The pre-trained model weights are made public, often along with the architecture. However, the training data or full training code might not be openly available.",
        "what_is_open": ["Model architecture", "Model weights"],
        "what_you_can_do": ["Use the pre-trained model for inference", "Fine-tune on custom data", "Adapt for specific tasks", "Deploy for various applications"],
        "what_you_cannot_do": ["Replicate the original training (without original data/code)", "Fully understand training data biases (if not provided)", "Restrictions might apply (e.g., commercial use, MAU caps, export)"],
    },
    "Architecture Only": {
        "definition": "Only the architectural design of the model is public, but the trained weights and training data are not released. Users must train it from scratch.",
        "what_is_open": ["Model architecture (e.g., paper description, code defining layers)"],
        "what_you_can_do": ["Understand the model's design", "Implement the architecture", "Train a new model from scratch using the same architecture"],
        "what_you_cannot_do": ["Use a pre-trained model", "Fine-tune (as there are no weights)", "Deploy for immediate use without extensive training"],
    },
}
open_source_levels
```

```python
comparison_table = f"""| Openness level | What's open? | Impact on retraining/modifying |
| --- | --- | --- |
| Fully Open | {', '.join(open_source_levels['Fully Open']['what_is_open'])} | Allows full replication of training, extensive fine-tuning, adaptation for specific domains, inspection for biases, and commercial use (depending on license). Few restrictions beyond license compliance. |
| Weights Released | {', '.join(open_source_levels['Weights Released']['what_is_open'])} | Enables inference and fine-tuning on custom data. Cannot fully replicate original training or understand training data biases if not provided. Restrictions might apply (e.g., commercial use, MAU caps). |
| Architecture Only | {', '.join(open_source_levels['Architecture Only']['what_is_open'])} | Requires training a new model from scratch; no pre-trained weights for fine-tuning or immediate use. |
"""
print(comparison_table)
```

```python
comparative_paragraph = (
    "The 'Fully Open' level offers the most transparency and flexibility, providing access to all components from training data to code, enabling full replication and extensive fine-tuning. "
    "In contrast, 'Weights Released' models provide pre-trained weights and architecture, allowing fine-tuning and deployment, but limit the ability to understand or replicate original training data or processes. "
    "The 'Architecture Only' level is the least open, providing only the model's design, which necessitates training from scratch, offering no pre-trained utility. "
    "Each level presents a different trade-off between control, transparency, and immediate usability for developers and researchers."
)
print(comparative_paragraph)
```

```python
healthcare_prompt_answer = (
    "For retraining on sensitive clinical data, a 'Fully Open' model is most suitable due to its complete transparency, allowing thorough inspection for biases and ensuring compliance with regulatory standards. "
    "This level of openness is critical for adapting the model responsibly and effectively within a healthcare context."
)
print(healthcare_prompt_answer)
```
