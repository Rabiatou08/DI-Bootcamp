## Exercise 2: Multi-Part Prompt for Quiz Generation
**Scenario**: Create an educational prompt for 7th graders about volcanic eruptions.

### Tasks
- Ask for: 2-bullet summary + 3 MCQs (1 correct, 2 distractors).
- Require friendly tone, simple vocab (11?13 y/o), and Google Slides-ready output.
- Explain 3 ways your prompt improves on: `Make a quiz for kids about this article.`

```python
# TODO: write the full multi-part prompt
quiz_prompt = """
You are an experienced 5th-grade Middle School Science Teacher. Your goal is to convert the provided text about volcanic eruptions into highly engaging classroom material.

Follow this multi-part instruction pipeline step by step:

Part 1: Key Summary
- Summarize the core scientific concepts of the article in exactly 2 clear, punchy bullet points.

Part 2: Educational Quiz
- Generate 3 Multiple-Choice Questions (MCQs) based strictly on the summary and text.
- For each question, provide exactly 1 correct answer and 2 plausible distractors (wrong answers).
- Label the options clearly (A, B, C) and indicate the correct answer below each question.

Part 3: Target Audience & Formatting Constraints
- Tone & Vocabulary: Friendly, enthusiastic, and simple. Tailor all language to children aged 11 to 13. Avoid complex jargon unless explained simply.
- Output Format: Format the final response as a clean JSON block or structured text compatible with Google Slides copy-pasting. Use the following schema:
  Slide 1: [Title]
  Slide 2: [Summary Points]
  Slide 3: [Question 1 + Options]
  Slide 4: [Question 2 + Options]
  Slide 5: [Question 3 + Options]

Article Text to process:
[INSERT ARTICLE TEXT HERE]
"""
print("Quiz Prompt Registered.")
```

```python
# TODO: explain three improvements over the vague prompt
improvements = [
    "Explicitly defines the target age group (11-13 years old) to prevent vocabulary that is either too childish or too academic.",
    "Enforces a specific structural layout (2 bullets, 3 MCQs with 1 right / 2 wrong options) to prevent the AI from generating random question formats.",
    "Demands a precise output schema compatible with Google Slides, eliminating raw blocks of unstructured prose."
]
print("\nKey Improvements:\n", improvements)
```

### Tasks

```python
# Selection: Functional Style
functional_support_prompt = """
You are a precise Customer Support Automation System for an e-commerce platform. Your sole task is to ingest raw, emotional customer emails and transform them into structured technical tickets for our support team.

Instructions:
1. Extract the customer's Core Issue (e.g., delayed shipping, defective item).
2. Extract the Order ID (if mentioned).
3. Classify the customer's sentiment into one of three categories: [HIGH URGENCY], [MEDIUM URGENCY], [LOW URGENCY].
4. Rewrite the customer's complaint into a polite, neutral, 2-sentence summary for the internal team.

Constraints:
- Output only valid JSON format containing the keys: "issue", "order_id", "urgency", "summary". Do not include any conversational preamble or pleasantries.

Customer Input:
[INSERT CUSTOMER EMAIL HERE]
"""

features = [
    "Zero conversational filler: The prompt uses strict negative constraints ('Do not include any preamble') to focus entirely on data transformation.",
    "Structured data mapping: It explicitly dictates the output schema (JSON with 4 specific keys) to allow direct software parsing."
]

justification = "The Functional style is superior for customer support ticketing because automation scripts require predictable, machine-readable data layouts. Conversational or exploratory fluff would break backend parsers and delay ticket routing."

print("Features:", features)
print("Justification:", justification)
```
