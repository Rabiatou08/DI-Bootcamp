Mini-projet : Analyse de données pour la stratégie marketing

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Charger le dataset
url = "https://github.com/devtlv/Datasets-DA-Bootcamp-2-/raw/refs/heads/main/Week%205%20-%20Data%20Processing/W5D5%20-%20Mini-project%20-%20bis/US%20Superstore%20data.xls"

df = pd.read_excel(url)
display(df.head())
print(df.shape)
print(df.columns.tolist())

# Vérifier les valeurs manquantes
print("=== Valeurs manquantes ===")
print(df.isnull().sum())

# Vérifier les types
print("\n=== Types de colonnes ===")
print(df.dtypes)

# Convertir les dates
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])

# Extraire l'année et le mois
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month

print("\n=== Aperçu après prétraitement ===")
display(df.head())


# Quels États enregistrent le plus de ventes ?
plt.figure(figsize=(14, 6))

top_states = df.groupby('State')['Sales'].sum().sort_values(ascending=False).head(10)

sns.barplot(x=top_states.index, y=top_states.values, palette='Blues_d')
plt.title('Top 10 États par chiffre d\'affaires')
plt.xlabel('État')
plt.ylabel('Ventes ($)')
plt.xticks(rotation=45)
plt.show()

print("=== Top 10 États ===")
print(top_states)

# Filtrer New York et Californie
ny_ca = df[df['State'].isin(['New York', 'California'])]

# Ventes et bénéfices par état
comparison = ny_ca.groupby('State')[['Sales', 'Profit']].sum()
print("=== New York vs Californie ===")
print(comparison)

# Visualisation
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

comparison['Sales'].plot(
    kind='bar',
    ax=axes[0],
    color=['blue', 'orange'],
    title='Ventes totales'
)
axes[0].set_ylabel('Ventes ($)')
axes[0].set_xlabel('')

comparison['Profit'].plot(
    kind='bar',
    ax=axes[1],
    color=['blue', 'orange'],
    title='Bénéfices totaux'
)
axes[1].set_ylabel('Bénéfices ($)')
axes[1].set_xlabel('')

plt.tight_layout()
plt.show()

# Filtrer New York
ny = df[df['State'] == 'New York']

# Top clients par ventes
top_clients_ny = ny.groupby('Customer Name')['Sales'].sum().sort_values(ascending=False).head(10)

print("=== Top 10 clients New York ===")
print(top_clients_ny)

# Visualisation
plt.figure(figsize=(12, 5))
sns.barplot(x=top_clients_ny.values, y=top_clients_ny.index, palette='Greens_d')
plt.title('Top 10 clients New York par ventes')
plt.xlabel('Ventes ($)')
plt.ylabel('Client')
plt.show()

# Client exceptionnel
best_client = top_clients_ny.index[0]
best_sales = top_clients_ny.values[0]
print(f"\n Client exceptionnel : {best_client} avec {best_sales:.2f}$")

# Rentabilité par état
profit_state = df.groupby('State')['Profit'].sum().sort_values(ascending=False)

# Top 10 et Bottom 10
top10 = profit_state.head(10)
bottom10 = profit_state.tail(10)

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# États les plus rentables
sns.barplot(x=top10.values, y=top10.index, palette='Greens_d', ax=axes[0])
axes[0].set_title('Top 10 États les plus rentables')
axes[0].set_xlabel('Bénéfice ($)')

# États les moins rentables
sns.barplot(x=bottom10.values, y=bottom10.index, palette='Reds_d', ax=axes[1])
axes[1].set_title('Top 10 États les moins rentables')
axes[1].set_xlabel('Bénéfice ($)')

plt.tight_layout()
plt.show()

print("=== États les plus rentables ===")
print(top10)
print("\n=== États les moins rentables ===")
print(bottom10)

# Calculer bénéfices par client
profit_client = df.groupby('Customer Name')['Profit'].sum().sort_values(ascending=False)

# Calculer le cumulatif
cumsum = profit_client.cumsum()
total = profit_client.sum()
cumsum_pct = cumsum / total * 100

# Trouver le seuil 80%
idx_80 = (cumsum_pct >= 80).idxmax()
n_clients_80 = list(profit_client.index).index(idx_80) + 1
pct_clients = n_clients_80 / len(profit_client) * 100

print(f"=== Principe de Pareto ===")
print(f"Total clients : {len(profit_client)}")
print(f"Clients pour 80% des bénéfices : {n_clients_80}")
print(f"Pourcentage de clients : {pct_clients:.1f}%")

# Visualisation courbe de Pareto
plt.figure(figsize=(12, 5))
plt.plot(range(len(cumsum_pct)), cumsum_pct.values, color='blue')
plt.axhline(y=80, color='red', linestyle='--', label='80%')
plt.axvline(x=n_clients_80, color='green', linestyle='--', label=f'{n_clients_80} clients')
plt.title('Courbe de Pareto - Clients vs Bénéfices')
plt.xlabel('Nombre de clients')
plt.ylabel('% cumulatif des bénéfices')
plt.legend()
plt.show()

# Top 20 villes par ventes et bénéfices
top20_ventes = df.groupby('City')['Sales'].sum().sort_values(ascending=False).head(20)
top20_profits = df.groupby('City')['Profit'].sum().sort_values(ascending=False).head(20)

fig, axes = plt.subplots(2, 1, figsize=(14, 12))

# Top 20 villes par ventes
sns.barplot(x=top20_ventes.values, y=top20_ventes.index,
            palette='Blues_d', ax=axes[0])
axes[0].set_title('Top 20 villes par ventes')
axes[0].set_xlabel('Ventes ($)')

# Top 20 villes par bénéfices
sns.barplot(x=top20_profits.values, y=top20_profits.index,
            palette='Greens_d', ax=axes[1])
axes[1].set_title('Top 20 villes par bénéfices')
axes[1].set_xlabel('Bénéfices ($)')

plt.tight_layout()
plt.show()

# Différences de rentabilité
print("=== Top 20 villes - Ventes ===")
print(top20_ventes)
print("\n=== Top 20 villes - Bénéfices ===")
print(top20_profits)

# Top 20 clients par ventes
top20_clients = df.groupby('Customer Name')['Sales'].sum().sort_values(ascending=False).head(20)

plt.figure(figsize=(12, 6))
sns.barplot(x=top20_clients.values, y=top20_clients.index, palette='Oranges_d')
plt.title('Top 20 clients par ventes')
plt.xlabel('Ventes ($)')
plt.ylabel('Client')
plt.show()

# Courbe cumulative Pareto - Ventes
ventes_client = df.groupby('Customer Name')['Sales'].sum().sort_values(ascending=False)
cumsum_ventes = ventes_client.cumsum()
total_ventes = ventes_client.sum()
cumsum_pct_ventes = cumsum_ventes / total_ventes * 100

# Seuil 80%
idx_80 = (cumsum_pct_ventes >= 80).idxmax()
n_clients_80 = list(ventes_client.index).index(idx_80) + 1
pct_clients = n_clients_80 / len(ventes_client) * 100

print(f"=== Pareto Ventes ===")
print(f"Clients pour 80% des ventes : {n_clients_80}")
print(f"Pourcentage de clients : {pct_clients:.1f}%")

# Visualisation
plt.figure(figsize=(12, 5))
plt.plot(range(len(cumsum_pct_ventes)), cumsum_pct_ventes.values, color='orange')
plt.axhline(y=80, color='red', linestyle='--', label='80%')
plt.axvline(x=n_clients_80, color='green', linestyle='--', label=f'{n_clients_80} clients')
plt.title('Courbe de Pareto - Clients vs Ventes')
plt.xlabel('Nombre de clients')
plt.ylabel('% cumulatif des ventes')
plt.legend()
plt.show()
```

## Conclusion et Recommandations Marketing

### 1. États prioritaires
- Californie et New York sont les États avec le plus de ventes
- Concentrer les efforts marketing sur ces États

### 2. New York vs Californie
- Les deux États sont très performants
- Analyser les différences pour adapter les stratégies

### 3. Client exceptionnel New York
- Identifier et fidéliser ce client VIP
- Proposer des offres personnalisées

### 4. Principe de Pareto
- ~20% des clients génèrent 80% des bénéfices
- Concentrer les ressources sur ces clients clés

### 5. Top 20 villes
- Prioriser les villes avec fort CA ET fort bénéfice
- Éviter les villes avec fort CA mais faible bénéfice

### 6. Recommandations
- Fidéliser les top clients
- Investir dans les États/villes les plus rentables
- Réduire les remises dans les zones peu rentables
