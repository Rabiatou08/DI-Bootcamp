 Exercice 3 : Comprendre les distributions

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Définir la moyenne et l'écart type
moyenne = 50
ecart_type = 10

# Générer les valeurs de x
x = np.linspace(10, 90, 100)

# Calculer la distribution normale
y = norm.pdf(x, moyenne, ecart_type)

# Représentation graphique
plt.plot(x, y)

# Ajouter un titre et des labels
plt.title("Distribution normale (moyenne=50, écart type=10)")
plt.xlabel("Valeurs")
plt.ylabel("Densité de probabilité")

# Afficher le graphique
plt.show()

```

 Exercice 4 : Application du test t

```python
import numpy as np
from scipy.stats import ttest_ind

# Générer deux ensembles de données aléatoires
data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

# Effectuer le test t
t_stat, p_value = ttest_ind(data1, data2)

# Afficher les résultats
print("Statistique t :", t_stat)
print("Valeur p :", p_value)

# Interprétation
if p_value < 0.05:
    print("Les deux ensembles de données sont significativement différents.")
else:
    print("Il n'y a pas de différence significative entre les deux ensembles de données.")
```

3. Interprétation de la pente

La pente représente l’augmentation moyenne du prix d’une maison pour chaque mètre carré supplémentaire.

```python
import numpy as np
from scipy.stats import linregress

# Données
house_sizes = np.array([50, 70, 80, 100, 120])
house_prices = np.array([150000, 200000, 210000, 250000, 280000])

# Régression linéaire
result = linregress(house_sizes, house_prices)

# Récupérer les résultats
slope = result.slope
intercept = result.intercept

# Prédiction pour une maison de 90 m²
predicted_price = slope * 90 + intercept

# Affichage
print("Pente :", slope)
print("Ordonnée à l'origine :", intercept)
print("Prix estimé pour 90 m² :", predicted_price)
```

 Exercice 6 : Comprendre l’ANOVA

```python
from scipy.stats import f_oneway

# Données
fertilizer_1 = [5, 6, 7, 6, 5]
fertilizer_2 = [7, 8, 7, 9, 8]
fertilizer_3 = [4, 5, 4, 3, 4]

# Test ANOVA
F_statistic, p_value = f_oneway(
    fertilizer_1,
    fertilizer_2,
    fertilizer_3
)

# Affichage des résultats
print("Valeur F :", F_statistic)
print("Valeur p :", p_value)

# Interprétation
if p_value < 0.05:
    print("Les engrais ont des effets significativement différents.")
else:
    print("Les engrais n'ont pas d'effets significativement différents.")
```

Exercice 7 : Distributions de probabilité (facultatif)

```python
from scipy.stats import binom

# Paramètres
n = 10      # nombre de lancers
p = 0.5     # probabilité d'obtenir face
k = 5       # nombre exact de faces

# Calcul de la probabilité
probability = binom.pmf(k, n, p)

# Résultat
print("Probabilité d'obtenir exactement 5 faces :", probability)
```

 Exercice 8 : Coefficients de corrélation (facultatif)

```python
import pandas as pd
from scipy.stats import pearsonr, spearmanr

# Créer le dataset
data = pd.DataFrame({
    'age': [23, 25, 30, 35, 40],
    'income': [35000, 40000, 50000, 60000, 70000]
})

# Corrélation de Pearson
pearson_corr, pearson_p = pearsonr(
    data['age'],
    data['income']
)

# Corrélation de Spearman
spearman_corr, spearman_p = spearmanr(
    data['age'],
    data['income']
)

# Afficher les résultats
print("Corrélation de Pearson :", pearson_corr)
print("P-value Pearson :", pearson_p)

print("Corrélation de Spearman :", spearman_corr)
print("P-value Spearman :", spearman_p)
```
